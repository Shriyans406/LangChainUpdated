```python
import os
from langchain_groq import ChatGroq

model=ChatGroq(model="qwen/qwen3-12b")
response=model.invoke("Why do parrots talk?")
response
```

```python
model.invoke("Write a 200 words paragraph on AI")
```

```python
for chunk in model.stream("Write a 200 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("Why parrots speak?"):
    print(chunk.text, end="|", flush=True)
```

```python
responses=model.batch([
    "Why do parrots speak?",
    "How do airplanes fly?",
    "What is quantum computing?"

])
for response in responses:
    print(response)
```
